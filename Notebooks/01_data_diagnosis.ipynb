```python
# Show column names, data types, and non-null counts for Home Credit
# Useful first check: confirms row/column count and flags which columns
# are numeric vs object (categorical/string) before we look at missingness
df_home.info()
```

```python
# Count missing values per column, sorted from most missing to least
# With 122 columns this output will be long - that's expected for Home Credit
# We'll use this to decide which columns to drop (>20% missing, per our plan)
# and which need imputation
df_home.isnull().sum().sort_values(ascending=False)
```

```python
# Calculate missingness as a percentage of total rows, not just raw counts
# This lets us directly apply the 20% threshold from our Data Quality Assurance Plan
missing_pct = (df_home.isnull().sum() / len(df_home)) * 100
missing_pct = missing_pct.sort_values(ascending=False)

# Show only columns that actually have any missing values
missing_pct[missing_pct > 0]
```

```python
# How many columns exceed 20% missing (candidates to drop, per our plan)
cols_above_threshold = missing_pct[missing_pct > 20]
print(f"Columns above 20% missing: {len(cols_above_threshold)}")
print(cols_above_threshold)

# How many columns have some missingness but are below 20% (candidates for imputation)
cols_below_threshold = missing_pct[(missing_pct > 0) & (missing_pct <= 20)]
print(f"\nColumns with missing values but below 20%: {len(cols_below_threshold)}")
print(cols_below_threshold)
```

```python
# Check if AMT_REQ_CREDIT_BUREAU_* columns are missing together (same rows)
# If they are, this points to "no bureau inquiry record exists" rather than random gaps
bureau_cols = [
    "AMT_REQ_CREDIT_BUREAU_HOUR", "AMT_REQ_CREDIT_BUREAU_DAY",
    "AMT_REQ_CREDIT_BUREAU_WEEK", "AMT_REQ_CREDIT_BUREAU_MON",
    "AMT_REQ_CREDIT_BUREAU_QRT", "AMT_REQ_CREDIT_BUREAU_YEAR"
]

# True if a row is missing across ALL of these columns at once
same_rows_missing = df_home[bureau_cols].isnull().all(axis=1).sum()
print(f"Rows missing across ALL bureau columns simultaneously: {same_rows_missing}")
print(f"Rows missing in at least one bureau column: {df_home[bureau_cols].isnull().any(axis=1).sum()}")
```

```python
# Check if OWN_CAR_AGE is missing specifically for applicants who don't own a car
# FLAG_OWN_CAR should be 'N' for anyone missing OWN_CAR_AGE, if our theory is right
pd.crosstab(df_home["OWN_CAR_AGE"].isnull(), df_home["FLAG_OWN_CAR"])
```

```python
# Check whether OCCUPATION_TYPE missingness lines up with NAME_INCOME_TYPE
# (e.g. pensioners/unemployed people would have no occupation to report)
pd.crosstab(df_home["OCCUPATION_TYPE"].isnull(), df_home["NAME_INCOME_TYPE"])
```

```python
# Columns to drop: >20% missing, structural/property attributes
# OWN_CAR_AGE is deliberately excluded despite being >20% missing -
# we've confirmed its missingness is explained by FLAG_OWN_CAR == 'N' (MAR),
# so it's handled separately below rather than dropped
cols_to_drop = missing_pct[missing_pct > 20].index.tolist()
cols_to_drop.remove("OWN_CAR_AGE")

print(f"Dropping {len(cols_to_drop)} columns above 20% missing (OWN_CAR_AGE kept deliberately)")
df_home = df_home.drop(columns=cols_to_drop)
print(f"New shape: {df_home.shape}")
```

```python
# For non-car-owners, missing OWN_CAR_AGE means "not applicable" -> impute 0
df_home.loc[df_home["FLAG_OWN_CAR"] == "N", "OWN_CAR_AGE"] = 0

# For the small number of car owners still missing OWN_CAR_AGE (genuine gaps),
# impute using the median age among actual car owners only
car_owner_median_age = df_home.loc[df_home["FLAG_OWN_CAR"] == "Y", "OWN_CAR_AGE"].median()
df_home.loc[
    (df_home["FLAG_OWN_CAR"] == "Y") & (df_home["OWN_CAR_AGE"].isnull()),
    "OWN_CAR_AGE"
] = car_owner_median_age

# Confirm no missing values remain
print(df_home["OWN_CAR_AGE"].isnull().sum())
```

```python
# OCCUPATION_TYPE was dropped in Cell 7 - reload just that column from the
# original raw file and reattach it to df_home using the shared index,
# since row order has not changed
df_home_original = pd.read_csv(
    DATA_DIR / "Home Credit Default Risk Dataset" / "application_train.csv",
    usecols=["SK_ID_CURR", "OCCUPATION_TYPE"]
)

# Merge back on SK_ID_CURR (safer than relying on row index/order matching)
df_home = df_home.merge(df_home_original, on="SK_ID_CURR", how="left")

print(df_home.shape)  # should now show 74 columns (73 + OCCUPATION_TYPE back)
print(df_home["OCCUPATION_TYPE"].isnull().sum())  # should show ~96,391 missing
```

```python
# Rather than dropping OCCUPATION_TYPE, treat "no occupation recorded" as its
# own meaningful category rather than losing a variable likely relevant to
# credit risk (occupation relates to income stability)
df_home["OCCUPATION_TYPE"] = df_home["OCCUPATION_TYPE"].fillna("Unknown/Not Applicable")

# Confirm no missing values remain, and check the new category's size
print(df_home["OCCUPATION_TYPE"].isnull().sum())
print(df_home["OCCUPATION_TYPE"].value_counts())
```

```python
# AMT_REQ_CREDIT_BUREAU_* columns (13.5% missing, confirmed missing together
# in Cell 5 - representing "no bureau inquiry history"). Impute 0, since
# absence of a recorded inquiry count most plausibly means zero inquiries
bureau_cols = [
    "AMT_REQ_CREDIT_BUREAU_HOUR", "AMT_REQ_CREDIT_BUREAU_DAY",
    "AMT_REQ_CREDIT_BUREAU_WEEK", "AMT_REQ_CREDIT_BUREAU_MON",
    "AMT_REQ_CREDIT_BUREAU_QRT", "AMT_REQ_CREDIT_BUREAU_YEAR"
]
df_home[bureau_cols] = df_home[bureau_cols].fillna(0)

print(df_home[bureau_cols].isnull().sum())
```

```python
# EXT_SOURCE_2 (0.2% missing) and EXT_SOURCE_3 (19.8% missing) are external
# normalized credit scores - no clear structural "not applicable" logic here,
# so apply standard median imputation per our Data Quality Assurance Plan
df_home["EXT_SOURCE_2"] = df_home["EXT_SOURCE_2"].fillna(df_home["EXT_SOURCE_2"].median())
df_home["EXT_SOURCE_3"] = df_home["EXT_SOURCE_3"].fillna(df_home["EXT_SOURCE_3"].median())

print(df_home[["EXT_SOURCE_2", "EXT_SOURCE_3"]].isnull().sum())
```

```python
# Handle the last handful of near-zero missingness columns:
# numeric columns -> median, categorical columns -> mode
remaining_missing = df_home.columns[df_home.isnull().sum() > 0]
print("Columns still with missing values:", remaining_missing.tolist())

for col in remaining_missing:
    if df_home[col].dtype in ["float64", "int64"]:
        df_home[col] = df_home[col].fillna(df_home[col].median())
    else:
        df_home[col] = df_home[col].fillna(df_home[col].mode()[0])

# Final confirmation - should be zero everywhere now
print(df_home.isnull().sum().sum())
```

```python
# Save the cleaned Home Credit dataset locally
HOME_DIR = DATA_DIR / "Home Credit Default Risk Dataset"
df_home.to_csv(HOME_DIR / "application_train_cleaned.csv", index=False)
print("Saved:", df_home.shape)
```

```python
# Calculate missingness as a percentage, same approach as Home Credit
missing_pct_lending = (df_lending.isnull().sum() / len(df_lending)) * 100
missing_pct_lending = missing_pct_lending.sort_values(ascending=False)
missing_pct_lending[missing_pct_lending > 0]
```

```python
# Check whether missing joint-income fields correspond to Individual applications
# (where joint fields simply wouldn't apply)
pd.crosstab(df_lending["annual_income_joint"].isnull(), df_lending["application_type"])
```

```python
# If delinq_2y (count of delinquencies in last 2 years) is 0 for most rows
# where months_since_last_delinq is missing, that supports "never happened"
# rather than "data unavailable"
pd.crosstab(df_lending["months_since_last_delinq"].isnull(), df_lending["delinq_2y"] == 0)
```

```python
# Joint-income fields are missing exactly when application_type is "individual" -
# confirmed 100% clean structural MAR. Impute 0 for numeric, "Not Applicable" for categorical,
# since these applicants simply have no joint co-borrower
df_lending["annual_income_joint"] = df_lending["annual_income_joint"].fillna(0)
df_lending["debt_to_income_joint"] = df_lending["debt_to_income_joint"].fillna(0)
df_lending["verification_income_joint"] = df_lending["verification_income_joint"].fillna("Not Applicable")

print(df_lending[["annual_income_joint", "debt_to_income_joint", "verification_income_joint"]].isnull().sum())
```

```python
# Missing months_since_last_delinq is confirmed to correspond overwhelmingly (99.8%)
# with delinq_2y == 0, i.e. no delinquency has ever occurred - not a real data gap.
# Impute with a large sentinel value (999) to represent "no delinquency on record",
# which keeps the column numeric and lets models learn this is functionally "very long ago / never"
df_lending["months_since_last_delinq"] = df_lending["months_since_last_delinq"].fillna(999)
df_lending["months_since_90d_late"] = df_lending["months_since_90d_late"].fillna(999)
df_lending["months_since_last_credit_inquiry"] = df_lending["months_since_last_credit_inquiry"].fillna(999)

print(df_lending[["months_since_last_delinq", "months_since_90d_late", "months_since_last_credit_inquiry"]].isnull().sum())
```

```python
# emp_title/emp_length: likely unemployed or non-disclosed - treat as its own category/value
df_lending["emp_title"] = df_lending["emp_title"].fillna("Not Disclosed")
df_lending["emp_length"] = df_lending["emp_length"].fillna(0)  # 0 years = plausible for unemployed/new

# Remaining numeric gaps: standard median imputation per our plan
df_lending["num_accounts_120d_past_due"] = df_lending["num_accounts_120d_past_due"].fillna(
    df_lending["num_accounts_120d_past_due"].median())
df_lending["debt_to_income"] = df_lending["debt_to_income"].fillna(df_lending["debt_to_income"].median())

# Final check - should be zero
print(df_lending.isnull().sum().sum())
```

```python
# Save the cleaned LendingClub dataset locally
LENDING_DIR = DATA_DIR / "Lending Club Loan Dataset"
df_lending.to_csv(LENDING_DIR / "loans_full_schema_cleaned.csv", index=False)
print("Saved:", df_lending.shape)
```
